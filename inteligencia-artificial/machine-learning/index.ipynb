```python
from sklearn.datasets import load_iris, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier, MLPRegressor
from sklearn.metrics import accuracy_score, mean_squared_error

# Exemplo 1: classificação de flores (Iris)
iris = load_iris()
X_treino, X_teste, y_treino, y_teste = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42, stratify=iris.target
)

classificador = make_pipeline(
    StandardScaler(),
    MLPClassifier(hidden_layer_sizes=(10,), max_iter=1500, random_state=42)
)
classificador.fit(X_treino, y_treino)

previsoes = classificador.predict(X_teste)
print("Acurácia na classificação:", accuracy_score(y_teste, previsoes))

# Exemplo 2: regressão para estimar a progressão do diabetes
diabetes = load_diabetes()
X_treino, X_teste, y_treino, y_teste = train_test_split(
    diabetes.data, diabetes.target, test_size=0.2, random_state=42
)

regressor = make_pipeline(
    StandardScaler(),
    MLPRegressor(hidden_layer_sizes=(32, 16), max_iter=2000, random_state=42)
)
regressor.fit(X_treino, y_treino)

previsoes = regressor.predict(X_teste)
rmse = mean_squared_error(y_teste, previsoes) ** 0.5
print("Erro médio (RMSE) na regressão:", rmse)
```
